```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Read the data and skip the first row
df = pd.read_csv('/Users/ellayoung/Desktop/metabolo_confi_score/data/base_data_202501161036.txt', sep='|', skipinitialspace=True)
df = df.iloc[1:]  # Skip the first row

# Clean column names and data
df.columns = df.columns.str.strip()
for col in df.columns:
    df[col] = df[col].str.strip() if df[col].dtype == 'object' else df[col]

def clean_curly_list(s):
    try:
        values = s.strip('{}').split(',')
        return [float(x.strip()) for x in values if x.strip()]
    except Exception as e:
        print(f"Error processing string: {s}")
        print(f"Error: {e}")
        return []

# Get column names
ri_col = [col for col in df.columns if 'ri_list' in col][0]
mass_col = [col for col in df.columns if 'mass_list' in col][0]
```

```python
import pandas as pd
import numpy as np

# Convert 'mass_list' and 'ri_list' from string to numerical lists
def convert_set_to_array(value):
    try:
        # If the value is a set, convert it to a sorted list and then to a NumPy array
        if isinstance(value, set):
            return np.array(sorted(value), dtype=float)
        elif isinstance(value, str):
            # Convert string representation of a set to a Python set, then to a NumPy array
            return np.array(sorted(eval(value)), dtype=float)
        else:
            return np.array([], dtype=float)  # Default to empty array for invalid entries
    except Exception as e:
        print(f"Error converting value: {value} -> {e}")
        return np.array([], dtype=float)

# Apply the conversion to 'mass_list' and 'ri_list'
df['mass_list'] = df['mass_list'].apply(convert_set_to_array)
df['ri_list'] = df['ri_list'].apply(convert_set_to_array)


# Calculate statistics for each internal standard
df['mass_mean'] = df['mass_list'].apply(np.mean)
df['mass_std'] = df['mass_list'].apply(np.std)
df['ri_mean'] = df['ri_list'].apply(np.mean)
df['ri_std'] = df['ri_list'].apply(np.std)

# Aggregate statistics across all internal standards
mass_std_global_mean = df['mass_std'].mean()
mass_std_global_std = df['mass_std'].std()

ri_std_global_mean = df['ri_std'].mean()
ri_std_global_std = df['ri_std'].std()

# Output the results
print(f"Global m/z Std Dev: Mean = {mass_std_global_mean:.6f}, Std Dev = {mass_std_global_std:.6f}")
print(f"Global RI Std Dev: Mean = {ri_std_global_mean:.6f}, Std Dev = {ri_std_global_std:.6f}")

# Define global tolerance ranges
mass_tolerance_range = (mass_std_global_mean - mass_std_global_std, 
                        mass_std_global_mean + mass_std_global_std)
ri_tolerance_range = (ri_std_global_mean - ri_std_global_std, 
                      ri_std_global_mean + ri_std_global_std)

print(f"m/z Tolerance Range: {mass_tolerance_range}")
print(f"RI Tolerance Range: {ri_tolerance_range}")
```

```python
# Calculate deltas from mean for each standard
results = []

for idx, row in df.iterrows():
    compound_id = row['compound_splash'].strip()
    ri_values = clean_curly_list(row[ri_col])
    mass_values = clean_curly_list(row[mass_col])
    
    if len(ri_values) > 0 and len(mass_values) > 0:
        # Calculate mean values as reference points
        ri_mean = np.mean(ri_values)
        mass_mean = np.mean(mass_values)
        
        # Calculate deltas
        ri_deltas = [abs(ri - ri_mean) for ri in ri_values]
        mass_errors_ppm = [abs((mass - mass_mean)/mass_mean * 1e6) for mass in mass_values]
        
        # Store results
        results.append({
            'compound_id': compound_id,
            'rt_mean': ri_mean,
            'rt_std': np.std(ri_values),
            'rt_max_delta': max(ri_deltas),
            'mass_mean': mass_mean,
            'mass_std': np.std(mass_values),
            'mass_max_error_ppm': max(mass_errors_ppm)
        })

# Convert to DataFrame
results_df = pd.DataFrame(results)

# Plot distributions
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# RT deltas distribution
sns.histplot(data=results_df, x='rt_std', ax=ax1)
ax1.set_title('RT Standard Deviation Distribution')
ax1.set_xlabel('RT Standard Deviation')

# Mass errors distribution
sns.histplot(data=results_df, x='mass_max_error_ppm', ax=ax2)
ax2.set_title('Mass Error Distribution')
ax2.set_xlabel('Max Mass Error (ppm)')

plt.tight_layout()
plt.show()

# Calculate recommended tolerance windows (3 sigma)
rt_window = 3 * results_df['rt_std'].median()
mass_window = 3 * results_df['mass_max_error_ppm'].median()

print(f"\nRecommended tolerance windows based on standards:")
print(f"RT window: ±{rt_window:.3f}")
print(f"Mass window: ±{mass_window:.1f} ppm")

# Check for correlation between RT and mass errors
correlation = results_df['rt_std'].corr(results_df['mass_max_error_ppm'])
print(f"\nCorrelation between RT and mass errors: {correlation:.3f}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Function to calculate RT statistics for each compound
def analyze_rt_variation(df, rt_col='ri_list', min_rt=14):
    rt_stats = []
    
    for idx, row in df.iterrows():
        compound_id = row['compound_splash'].strip()
        rt_values = clean_curly_list(row[rt_col])
        
        # Filter RTs > min_rt
        valid_rts = [rt for rt in rt_values if rt > min_rt]
        
        if len(valid_rts) > 0:
            rt_stats.append({
                'compound_id': compound_id,
                'mean_rt': np.mean(valid_rts),
                'std_rt': np.std(valid_rts),
                'n_measurements': len(valid_rts),
                'rsd': (np.std(valid_rts) / np.mean(valid_rts)) * 100  # Relative standard deviation
            })
    
    return pd.DataFrame(rt_stats)

# Calculate RT statistics
rt_stats_df = analyze_rt_variation(df)

# Plot distributions
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Standard deviation distribution
sns.histplot(data=rt_stats_df, x='std_rt', ax=ax1)
ax1.set_title('RT Standard Deviation Distribution')
ax1.set_xlabel('RT Standard Deviation (s)')

# RSD distribution 
sns.histplot(data=rt_stats_df, x='rsd', ax=ax2)
ax2.set_title('RT Relative Standard Deviation Distribution')
ax2.set_xlabel('RSD (%)')

plt.tight_layout()
plt.show()

# Print summary statistics
print("\nRT Variation Statistics:")
print("\nStandard Deviation Summary:")
print(rt_stats_df['std_rt'].describe())
print("\nRelative Standard Deviation Summary:")
print(rt_stats_df['rsd'].describe())
```

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Read the data and skip the first row
df = pd.read_csv('data/base_data_202501161036.txt', sep='|', skipinitialspace=True)
df = df.iloc[1:]  # Skip the first row

# Clean column names and data
df.columns = df.columns.str.strip()
for col in df.columns:
    df[col] = df[col].str.strip() if df[col].dtype == 'object' else df[col]

def clean_curly_list(s):
    try:
        values = s.strip('{}').split(',')
        return [float(x.strip()) for x in values if x.strip()]
    except Exception as e:
        print(f"Error processing string: {s}")
        print(f"Error: {e}")
        return []

# Get column names
ri_col = [col for col in df.columns if 'ri_list' in col][0]
mass_col = [col for col in df.columns if 'mass_list' in col][0]



def bootstrap_analysis(data, n_bootstrap=1000):
    """
    Perform bootstrap analysis on a dataset to get confidence intervals
    
    Parameters:
    -----------
    data : list or array
        The measurements (either RT or mass values)
    n_bootstrap : int
        Number of bootstrap iterations
    
    Returns:
    --------
    dict containing:
        - confidence intervals for mean and standard deviation
        - bootstrap distributions for visualization
    """
    boot_means = []
    boot_stds = []
    
    # Perform bootstrap iterations
    for _ in range(n_bootstrap):
        # Sample with replacement
        boot_sample = np.random.choice(data, size=len(data), replace=True)
        
        # Calculate statistics for this bootstrap sample
        boot_means.append(np.mean(boot_sample))
        boot_stds.append(np.std(boot_sample))
    
    # Calculate confidence intervals
    ci_mean = np.percentile(boot_means, [2.5, 97.5])
    ci_std = np.percentile(boot_stds, [2.5, 97.5])
    
    return {
        'ci_mean': ci_mean,
        'ci_std': ci_std,
        'boot_means': boot_means,
        'boot_stds': boot_stds
    }

# Now let's modify your visualization code to include bootstrap results
def analyze_compound_with_bootstrap(ri_values, mass_values, compound_id):
    """
    Analyze compound measurements with bootstrap confidence intervals
    """
    if len(ri_values) > 0 and len(mass_values) > 0:
        # Perform bootstrap analysis
        ri_boot = bootstrap_analysis(ri_values)
        mass_boot = bootstrap_analysis(mass_values)
        
        # Create figure with histograms and bootstrap distributions
        fig = plt.figure(figsize=(15, 10))
        gs = plt.GridSpec(2, 2)
        
        # RT histogram
        ax1 = fig.add_subplot(gs[0, 0])
        sns.histplot(data=ri_values, kde=True, ax=ax1, bins=min(20, len(ri_values)))
        ax1.axvline(np.mean(ri_values), color='r', linestyle='--', label='Mean')
        ax1.axvspan(ri_boot['ci_mean'][0], ri_boot['ci_mean'][1], 
                   color='r', alpha=0.2, label='95% CI for mean')
        ax1.set_title(f'Retention Indices Distribution')
        ax1.set_xlabel('Retention Index')
        ax1.set_ylabel('Count')
        ax1.legend()
        
        # Mass histogram
        ax2 = fig.add_subplot(gs[0, 1])
        sns.histplot(data=mass_values, kde=True, ax=ax2, bins=min(20, len(mass_values)))
        ax2.axvline(np.mean(mass_values), color='r', linestyle='--', label='Mean')
        ax2.axvspan(mass_boot['ci_mean'][0], mass_boot['ci_mean'][1],
                   color='r', alpha=0.2, label='95% CI for mean')
        ax2.set_title(f'Masses Distribution')
        ax2.set_xlabel('Mass')
        ax2.set_ylabel('Count')
        ax2.legend()
        
        # Bootstrap distributions
        ax3 = fig.add_subplot(gs[1, 0])
        sns.histplot(data=ri_boot['boot_means'], kde=True, ax=ax3)
        ax3.set_title('Bootstrap Distribution of RT Mean')
        ax3.set_xlabel('Mean RT')
        
        ax4 = fig.add_subplot(gs[1, 1])
        sns.histplot(data=mass_boot['boot_means'], kde=True, ax=ax4)
        ax4.set_title('Bootstrap Distribution of Mass Mean')
        ax4.set_xlabel('Mean Mass')
        
        plt.suptitle(f'Compound Analysis: {compound_id}\nwith Bootstrap Confidence Intervals', y=1.02)
        plt.tight_layout()
        plt.show()
        
        # Print statistics with confidence intervals
        print("\nBootstrap Analysis Results:")
        print(f"\nRetention Index:")
        print(f"Mean: {np.mean(ri_values):.4f} (95% CI: [{ri_boot['ci_mean'][0]:.4f}, {ri_boot['ci_mean'][1]:.4f}])")
        print(f"Standard Deviation: {np.std(ri_values):.4f} (95% CI: [{ri_boot['ci_std'][0]:.4f}, {ri_boot['ci_std'][1]:.4f}])")
        
        print(f"\nMass:")
        print(f"Mean: {np.mean(mass_values):.4f} (95% CI: [{mass_boot['ci_mean'][0]:.4f}, {mass_boot['ci_mean'][1]:.4f}])")
        print(f"Standard Deviation: {np.std(mass_values):.4f} (95% CI: [{mass_boot['ci_std'][0]:.4f}, {mass_boot['ci_std'][1]:.4f}])")
        
        return ri_boot, mass_boot
    else:
        print("Not enough data to perform analysis")
        return None, None

# Modify your main loop to use the new analysis function
for idx, row in df.iterrows():
    compound_id = row['compound_splash'].strip()
    ri_values = clean_curly_list(row[ri_col])
    mass_values = clean_curly_list(row[mass_col])
    
    print(f"\nAnalyzing compound {idx}: {compound_id}")
    print(f"Number of measurements - RI: {len(ri_values)}, Mass: {len(mass_values)}")
    
    ri_boot, mass_boot = analyze_compound_with_bootstrap(ri_values, mass_values, compound_id)
    
    # Optional: add a pause between plots
    if idx < max(df.index):
        input("Press Enter to continue to next compound...")
```

```python
def calculate_rt_match_score(experimental_data, reference_data):
    """
    Calculate retention time match score between experimental and reference data
    
    Parameters:
    -----------
    experimental_data : dict
        Contains:
        - rt_mean: mean retention time from bin
        - rt_ci: confidence interval for RT from bootstrap
        - rt_std: standard deviation of RT in bin
    
    reference_data : dict
        Contains:
        - rt: reference retention time from MassWiki
        - rt_tolerance: known tolerance for the method
    
    Returns:
    --------
    float : Score between 0 and 1, where 1 is perfect match
    """
    # 1. Calculate the normalized RT difference
    rt_difference = abs(experimental_data['rt_mean'] - reference_data['rt'])
    
    # 2. Consider the measurement uncertainty from bootstrap
    rt_uncertainty = experimental_data['rt_ci'][1] - experimental_data['rt_ci'][0]
    
    # 3. Consider method-specific tolerance
    total_tolerance = max(rt_uncertainty, reference_data['rt_tolerance'])
    
    # 4. Calculate score using Gaussian function
    # This gives 1.0 for perfect match, declining smoothly as difference increases
    score = np.exp(-(rt_difference**2) / (2 * total_tolerance**2))
    
    return {
        'score': score,
        'details': {
            'rt_difference': rt_difference,
            'uncertainty': rt_uncertainty,
            'tolerance': total_tolerance
        }
    }

# Example usage:
experimental_data = {
    'rt_mean': 24.12,        # From your bin statistics
    'rt_ci': [24.11, 24.13], # From bootstrap analysis
    'rt_std': 0.5886         # From your bin statistics
}

reference_data = {
    'rt': 24.15,            # From MassWiki
    'rt_tolerance': 0.1     # Method-specific tolerance
}

match_result = calculate_rt_match_score(experimental_data, reference_data)
```
